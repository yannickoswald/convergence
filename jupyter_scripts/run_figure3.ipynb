```python
# --- ADD THIS AFTER YOUR EXISTING CODE ---
from scipy.stats import linregress
import numpy as np
import matplotlib.pyplot as plt

# 1. Extract the data for the FIRST YEAR (2022) only
first_year_gdp_pc = []
first_year_carbon_pc = []

target_year = 2022 # The starting year of the simulation

# We use scenario_1, which is the final scenario calculated in your loop
for country_code, country in scenario_1.countries.items():
    # Extract the exact values for 2022
    gdp_val = country.gdppc_trajectory.get(target_year)
    carbon_val = country.carbon_emissions_pc_trajectory.get(target_year)
    
    # Ensure values are strictly positive to avoid issues with log transformation
    if gdp_val is not None and carbon_val is not None and gdp_val > 0 and carbon_val > 0:
        first_year_gdp_pc.append(gdp_val)
        first_year_carbon_pc.append(carbon_val)

# 2. Apply log-log transformation
log_gdp_pc = np.log(first_year_gdp_pc)
log_carbon_pc = np.log(first_year_carbon_pc)

# 3. Perform linear regression to find the elasticity (slope)
slope, intercept, r_value, p_value, std_err = linregress(log_gdp_pc, log_carbon_pc)
elasticity = slope

# 4. Create the plot
fig_elasticity, ax_elasticity = plt.subplots(figsize=(8, 6))

# Scatter plot of countries
ax_elasticity.scatter(log_gdp_pc, log_carbon_pc, alpha=0.6, edgecolors='w', s=50, label=f'Countries (Year {target_year})')

# Plot regression line
x_fit = np.linspace(min(log_gdp_pc), max(log_gdp_pc), 100)
y_fit = intercept + slope * x_fit
ax_elasticity.plot(x_fit, y_fit, color='red', linewidth=2, 
                   label=f'Fit (Elasticity $\\epsilon$ = {elasticity:.3f})\n$R^2$ = {r_value**2:.3f}')

# Formatting
ax_elasticity.set_xlabel('Log(GDP per capita)')
ax_elasticity.set_ylabel('Log(Carbon per capita)')
ax_elasticity.set_title(f'Cross-Country Income Elasticity of Carbon Emissions ({target_year})')
ax_elasticity.legend()
ax_elasticity.grid(True, linestyle='--', alpha=0.5)

# Save and show
plt.tight_layout()
plt.savefig(f"cross_country_elasticity_{target_year}.png", dpi=300)
plt.show()

# 5. Print out the measurements
print(f"Calculated Cross-Country Elasticity for {target_year}: {elasticity:.3f}")
print(f"R-squared: {r_value**2:.3f}")
print(f"Standard Error: {std_err:.3f}")
```

```python
from scipy.stats import linregress
import numpy as np
import matplotlib.pyplot as plt

# ==============================================================================
# POST-SIMULATION ANALYSIS: Cross-Country Elasticity Over Time
# ==============================================================================

# 1. Safely determine the exact years that were simulated
# We just look at the keys of the trajectory dictionary from any one country
sample_country = list(scenario_1.countries.values())[0]
simulated_years = sorted(list(sample_country.gdppc_trajectory.keys()))



elasticity_over_time = []
r_squared_over_time = []

# 2. Loop over every single year in the simulation
for year in simulated_years:
    gdp_pc_this_year = []
    carbon_pc_this_year = []
    
    # Extract the cross-sectional data (one data point per country) for this specific year
    for country in scenario_1.countries.values():
        gdp_val = country.gdppc_trajectory.get(year)
        carbon_val = country.carbon_emissions_pc_trajectory.get(year)
        
        # Ensure values exist and are strictly positive (required for log transformation)
        if gdp_val is not None and carbon_val is not None and gdp_val > 0 and carbon_val > 0:
            gdp_pc_this_year.append(gdp_val)
            carbon_pc_this_year.append(carbon_val)
            
    # 3. Calculate the cross-country elasticity for this year via log-log regression
    if len(gdp_pc_this_year) > 2:  # Need enough data points for a valid regression
        log_gdp = np.log(gdp_pc_this_year)
        log_carbon = np.log(carbon_pc_this_year)
        
        slope, intercept, r_value, p_value, std_err = linregress(log_gdp, log_carbon)
        
        elasticity_over_time.append(slope)
        r_squared_over_time.append(r_value**2)
    else:
        elasticity_over_time.append(np.nan)
        r_squared_over_time.append(np.nan)

# ==============================================================================
# PLOTTING THE ELASTICITY TIME-SERIES
# ==============================================================================

fig_el, ax_el = plt.subplots(figsize=(8, 5))

# Plot the elasticity curve
ax_el.plot(simulated_years, elasticity_over_time, color='teal', linewidth=2.5, 
           label="Cross-Country Elasticity ($\\epsilon$)")

# Add some helpful reference lines
ax_el.axhline(1.0, color='gray', linestyle='--', alpha=0.7, 
              label="1-to-1 Baseline ($\\epsilon = 1$)")
ax_el.axhline(0.0, color='red', linestyle=':', alpha=0.7, 
              label="Absolute Decoupling Threshold ($\\epsilon = 0$)")

# Formatting
ax_el.set_xlabel('Year', fontsize=12)
ax_el.set_ylabel('Income Elasticity of Carbon Emissions', fontsize=12)
ax_el.set_title('Evolution of Global Cross-Country Elasticity Over Time\n(Scenario: Income Goal ' + str(scenario_1.income_goal) + ')', fontsize=14)
ax_el.legend(loc='upper right', frameon=False)
ax_el.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()

# Save the figure
plt.savefig("cross_country_elasticity_timeseries.png", dpi=300)
plt.show()

# Optional: Save the time-series data to a CSV for your records
import pandas as pd
elasticity_df = pd.DataFrame({
    'Year': simulated_years,
    'Cross_Country_Elasticity': elasticity_over_time,
    'R_Squared': r_squared_over_time
})
elasticity_df.to_csv("cross_country_elasticity_over_time.csv", index=False)
```
